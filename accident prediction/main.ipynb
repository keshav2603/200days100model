```python
df.drop(columns=["id"],inplace=True)
```

```python
x_train, x_test,y_train, y_test = train_test_split(df.drop(columns=["accident_risk"]),df["accident_risk"],test_size=0.3)
```

```python
bool_cols = x_train.select_dtypes(include='bool')


for col in bool_cols:
    x_train[col] = x_train[col].astype(int)
    x_test[col] = x_test[col].astype(int)
```

```python
x_train
```

```python
obj_col = x_train.select_dtypes(include="object").columns

obj_col
```

```python
for col in obj_col:
    print(y_train.groupby(x_train[col]).mean())
```

as all the categorical colm will be onehotencoded able

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe=OneHotEncoder(sparse_output=False,drop='first')
```

```python
x_train_encoded=ohe.fit_transform(x_train[obj_col])
x_test_encoded=ohe.fit_transform(x_test[obj_col])
```

```python
x_train_encoded
```

```python
x_train_other=x_train.select_dtypes(include=["int","float"])
x_test_other=x_test.select_dtypes(include=["int","float"])
```

```python
x_train_other.values
```

```python
x_test_rest=x_test_other.values
x_train_rest=x_train_other.values
x_train_rest
```

```python
x_train_encoded
```

```python
final_train = np.hstack([x_train_rest, x_train_encoded])
final_test = np.hstack([x_test_rest, x_test_encoded])
```

```python
final_train
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
model=DecisionTreeRegressor()

model.fit(final_train,y_train)
```

```python
y_pred=model.predict(final_test)
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y_pred,y_test)
```
